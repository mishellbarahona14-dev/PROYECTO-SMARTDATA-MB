## **Grants**

```sql
-- ==============================================================================
-- 1. PERMISOS A NIVEL DE CATÁLOGO Y ESQUEMAS
-- ==============================================================================
-- Otorgar permiso de uso del catálogo a los grupos de trabajo
GRANT USE CATALOG ON CATALOG catalog_au TO `grupo_data_engineers`;
GRANT USE CATALOG ON CATALOG catalog_au TO `grupo_data_analysts`;
```

```sql
-- Otorgar permiso de uso de los esquemas según el rol
-- Los ingenieros acceden a todas las capas
GRANT USE SCHEMA ON SCHEMA catalog_au.bronze TO `grupo_data_engineers`;
GRANT USE SCHEMA ON SCHEMA catalog_au.silver TO `grupo_data_engineers`;
GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `grupo_data_engineers`;
```

```sql
-- Los analistas solo acceden a la capa final (Golden)
GRANT USE SCHEMA ON SCHEMA catalog_au.golden TO `grupo_data_analysts`;
```

```sql
-- ==============================================================================
-- 2. PERMISOS SOBRE LAS TABLAS BRONZE (INGESTA)
-- ==============================================================================
GRANT SELECT, MODIFY ON TABLE catalog_au.bronze.spotify_tracks TO `grupo_data_engineers`;
GRANT SELECT, MODIFY ON TABLE catalog_au.bronze.spotify_reviews TO `grupo_data_engineers`;
GRANT SELECT, MODIFY ON TABLE catalog_au.bronze.itunes_trends TO `grupo_data_engineers`;
```

```sql
-- ==============================================================================
-- 3. PERMISOS SOBRE LAS TABLAS SILVER (TRANSFORMACIÓN)
-- ==============================================================================
GRANT SELECT, MODIFY ON TABLE catalog_au.silver.music_trends_transformed TO `grupo_data_engineers`;
GRANT SELECT, MODIFY ON TABLE catalog_au.silver.spotify_reviews_clean TO `grupo_data_engineers`;
```

```sql
-- ==============================================================================
-- 4. PERMISOS SOBRE LAS TABLAS GOLDEN (NEGOCIO / POWER BI)
-- ==============================================================================
-- Permisos totales para ingenieros
GRANT SELECT, MODIFY ON TABLE catalog_au.golden.gold_music_metrics TO `grupo_data_engineers`;
GRANT SELECT, MODIFY ON TABLE catalog_au.golden.gold_reviews_metrics TO `grupo_data_engineers`;
```

```sql

-- Permisos de solo lectura para analistas y usuarios de negocio
GRANT SELECT ON TABLE catalog_au.golden.gold_music_metrics TO `grupo_data_analysts`;
GRANT SELECT ON TABLE catalog_au.golden.gold_reviews_metrics TO `grupo_data_analysts`;
GRANT SELECT ON TABLE catalog_au.golden.gold_music_metrics TO `mishellbarahona14@gmail.com`;
```
